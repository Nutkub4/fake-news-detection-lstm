# 1. Import Libraries

```python
import pickle
import numpy as np

from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

# 2. Load Trained Model

```python
MODEL_PATH = "../models/lstm_v2.keras"
model = load_model(MODEL_PATH)
print("Model loaded successfully")
```

# 3. Load Tokenizer

```python
MAX_LENGTH = 300
TOKENIZER_PATH = "../models/tokenizer.pkl"

with open(
    TOKENIZER_PATH,
    "rb"
) as file:

    tokenizer = pickle.load(file)

print("Tokenizer loaded successfully")
```

# 4. Create Prediction Function

```python
def predict_news(news_text):

    """
    Predict whether a news article is fake or real
    
    Output:
        prediction label
        confidence score
    """
    # Convert text to sequence
    sequence = tokenizer.texts_to_sequences([news_text])

    # Padding
    padded_sequence = pad_sequences(
        sequence,
        maxlen=MAX_LENGTH,
        padding="post",
        truncating="post"
    )

    # Prediction
    probability = model.predict(padded_sequence)[0][0]

    # Classification threshold

    if probability >= 0.5:
        label = "Real News"
        confidence = probability
    else:
        label = "Fake News"
        confidence = 1 - probability
        
    return label, confidence
```

# 5. Testing

```python
news_1 = """
The government announced a new economic policy
to support small businesses and increase employment.
"""

label, confidence = predict_news(news_1)

print("Prediction:", label)
print("Confidence:", round(confidence * 100, 2),"%")
```

```python
news_2 = """
A celebrity claims that drinking a special liquid
can make people live forever without any scientific proof.
"""

label, confidence = predict_news(news_2)

print("Prediction:", label)
print("Confidence:", round(confidence * 100, 2),"%")
```
